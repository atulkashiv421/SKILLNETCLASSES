# Create a Series:  
# Q1: Create a Pandas Series with values [10, 20, 30, 40, 50] and index [a, b, c, d, e].  

```python
import pandas as pd 
s=pd.Series([10,20,30,40,50],index=['a','b','c','d','e'])
print(s)
```

# Accessing Elements:  
  
# Q2:Access the third element of the Series created in question 1.  

```python
access=s['b':'d']
print(access)
```

# Operations on Series:  
  
# Q3: Add a Series with values [1, 2, 3, 4, 5] to the Series created in question 1. And multiply the series with 2.

```python
s2=[1,2,3,4,5]
add=s+s2
print(add)
multiply=s*s
print(multiply)
```

# Q4: Filter and print only the values greater than 30 from the series created in question 1.

```python
filterdata=(s>30)
print(filterdata)
```

# Missing Values:

# Q5: Create a new series with value [1,2,np.nan,4,5] and perform a forward fill to fill the missing value.

```python
import pandas as pd
import numpy as np
data=[1,2,np.nan,4,5]
a=pd.DataFrame(data)
print(a)
b=a.fillna(2)
print(b)
```

# Combining Series:

# Q10: Create two series with different values but some common indices. Concatenate them into single series.

```python
import pandas as pd
s1=pd.Series([1,2,3,4,5],index=['a','b','c','d','e'])
s2=pd.Series([6,7,8,9,10])
concatenate=pd.concat([s1,s2])
print(concatenate)
```
