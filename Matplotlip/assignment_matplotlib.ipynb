Follow along with these steps: **
- Create a figure object called fig using plt.figure() 
-  Use add_axes to add an axis to the figure canvas at [0,0,1,1]. Call this new axis.
- Plot (x,y) on that axes and set the labels and titles to match the plot below

```python
import matplotlib.pyplot as plt
import numpy as np
fig = plt.figure()
axes = fig.add_axes([0, 0, 1, 1])
x = [1, 2, 3, 4, 5]
y = [1, 2, 3, 4, 5]   
axes.plot(x, y)
axes.set_xlabel("x Axis")
axes.set_ylabel("y Axis")
axes.set_title("x vs y Plot")
plt.show()
```

# Question 3:
Now plot (x,y) on both axes. And call your figure object to show it

```python
import matplotlib.pyplot as plt
fig = plt.figure()
ax2 = fig.add_axes([0.2, 0.5, 0.2, 0.2])
x = [1, 2, 3, 4, 5]
y = [i**1 for i in x]
ax1.plot(x, y)
ax1.set_title("Main Plot (ax1)")
ax2.plot(x, y, "r")  
ax2.set_title("Inset Plot (ax2)", fontsize=8)

plt.show()
```

# Question 4: 
Now use x,y, and z arrays to recreate the plot below. Notice the xlimits and y limits on the inserted plot

```python
import matplotlib.pyplot as plt
import numpy as np
x = np.linspace(0, 10, 100)
y = x ** 1
z = x **2
fig = plt.figure()
ax1 = fig.add_axes([0, 0, 1, 1])
ax1.plot(x, y)
ax1.plot(x, z)
ax1.set_title("Main Plot")
ax1.set_xlabel("X Axis")
ax1.set_ylabel("Y / Z Axis")
ax1.legend()
ax2 = fig.add_axes([0.2, 0.5, 0.4, 0.4])
ax2.plot(x, y)
ax2.plot(x, z, "r")
ax2.set_xlim(0, 5)
ax2.set_ylim(0, 100)
ax2.set_title("Inset Zoom", fontsize=6)
plt.show()
```

# Question 5:
Use plt.subplots(nrows=1, ncols=2) to create the plot below.
Now plot (x,y) and (x,z) on the axes. Play around with the linewidth and style
1

```python
x = np.linspace(0, 10, 100)
y = x ** 1
z = x ** 2
fig = plt.figure()
ax1 = fig.add_axes([0, 0, 1, 1])
ax1.plot(x, y, color="g", linewidth=2, linestyle="--", label="y = x^2")
ax1.plot(x, z, color="m", linewidth=3, linestyle="-.", label="z = x^3")

ax1.set_title("Main Plot")
ax1.set_xlabel("X Axis")
ax1.set_ylabel("Y / Z Axis")
ax1.legend()
ax2 = fig.add_axes([0.2, 0.5, 0.4, 0.4])
ax2.plot(x, y, color="r", linewidth=1.5, linestyle=":")
ax2.plot(x, z, color="g", linewidth=2, linestyle="-")
plt.show()
```
